```python
text_file = '../data/train/text'
```

```python
counts = {}
with open(text_file, 'r') as f:
    for line in f:
        line = line.rstrip()
        id, text = line.split(' ', maxsplit=1)
        counts[text] = counts.get(text, 0) + 1
```

```python
import pandas as pd
```

```python
ser = pd.Series(counts)
```

```python
ser.sort_values(ascending=False, inplace=True)
```

```python
ser[ser >= 300]
```
